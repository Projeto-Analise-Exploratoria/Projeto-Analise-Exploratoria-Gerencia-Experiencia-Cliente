```python
import pandas as pd
import numpy as np

df1 = pd.read_csv("olist_orders_dataset.csv")
df2 = pd.read_csv("olist_order_items_dataset.csv")
df3 = pd.read_csv("olist_products_dataset.csv")
df4 = pd.read_csv("olist_order_reviews_dataset.csv")
df5 = pd.read_csv("olist_customers_dataset.csv")
```

**Lista de Ordens (DF1)**

```python
print(df1.info())
```

```python
df1.columns = ["id_pedido", "id_cliente", "status_pedido",
               "marca_temporal_compra_pedido", "pedido_aprovado_as",
               "data_entrega_pedido_transporadora",
               "data_entrega_pedido_cliente", "data_estimada_entrega_pedido"]
print(df1.head())
```

```python
print("\n---- Nulos por coluna ----")
print(df1.isna().sum())

print("\n---- Porcentagem de Nulos ----")
print(df1.isna().mean() * 100)
```

```python
colunas_com_nulos_df1 = ['pedido_aprovado_as','data_entrega_pedido_transporadora','data_entrega_pedido_cliente']
linhas_nulas_df1 = df1[df1[colunas_com_nulos_df1].isna().any(axis = 1)]
print(linhas_nulas_df1)
```

```python
df1.columns = df1.columns.str.strip().str.lower().str.replace(' ', '_')
```

As colunas `marca_temporal_compra_pedido`, `pedido_aprovado_as`, `data_entrega_pedido_transporadora`, `data_entrega_pedido_cliente` e `data_estimada_entrega_pedido` estão incorretamente classificadas como `object` ao invés de `datetime`.

```python
df1['marca_temporal_compra_pedido'] = pd.to_datetime(df1['marca_temporal_compra_pedido'], errors='coerce')
df1['pedido_aprovado_as'] = pd.to_datetime(df1['pedido_aprovado_as'], errors='coerce')
df1['data_entrega_pedido_transporadora'] = pd.to_datetime(df1['data_entrega_pedido_transporadora'], errors='coerce')
df1['data_entrega_pedido_cliente'] = pd.to_datetime(df1['data_entrega_pedido_cliente'], errors='coerce')
df1['data_estimada_entrega_pedido'] = pd.to_datetime(df1['data_estimada_entrega_pedido'], errors='coerce')
```

**Lista de produtos (DF3)**

```python
print(df3.info())
```

```python
df3.columns = ["id_produto", "nome_categoria_produto",
               "comprimento_nome_produto", "comprimento_descricao_produto",
               "qtd_fotos_produto", "peso_produto_g", "comprimento_produto_cm",
               "altura_produto_cm", "largura_produto_cm"]
print(df3.head())
```

```python
print("\n---- Nulos por coluna ----")
print(df3.isna().sum())

print("\n---- Porcentagem de Nulos ----")
print(df3.isna().mean() * 100)
```

Dataframe com nulos em quase todas as colunas. Quatro colunas tem 610 valores nulos e outras quatro possuem 2 valores nulos, indicando que as mesmas linhas possuem várias colunas nulas.


```python
colunas_com_nulos_df3 = ['nome_categoria_produto','comprimento_nome_produto',
                         'comprimento_descricao_produto', 'qtd_fotos_produto',
                         'peso_produto_g','comprimento_produto_cm',
                         'altura_produto_cm', 'largura_produto_cm']
linhas_nulas_df3 = df3[df3[colunas_com_nulos_df3].isna().any(axis = 1)]
print(linhas_nulas_df3)
```

As colunas `comprimento_nome_produto`, `comprimento_decricao_produto` e `qtd_fotos_produto` estão classificados como `float` sendo que seus valores são todos inteiros. Também preencher os valores nulos das colunas de forma apropriada.

```python
df3['nome_categoria_produto'] = df3['nome_categoria_produto'].fillna("Não classificado")
df3['comprimento_nome_produto'] = df3['comprimento_nome_produto'].fillna(df3['comprimento_nome_produto'].median())
df3['comprimento_descricao_produto'] = df3['comprimento_descricao_produto'].fillna(df3['comprimento_descricao_produto'].median())
df3['qtd_fotos_produto'] = df3['qtd_fotos_produto'].fillna(0)
df3['peso_produto_g'] = df3['peso_produto_g'].fillna(df3['peso_produto_g'].median())
df3['comprimento_produto_cm'] = df3['comprimento_produto_cm'].fillna(df3['comprimento_produto_cm'].median())
df3['altura_produto_cm'] = df3['altura_produto_cm'].fillna(df3['altura_produto_cm'].median())
df3['largura_produto_cm'] = df3['largura_produto_cm'].fillna(df3['largura_produto_cm'].median())
```

```python
df3['comprimento_nome_produto'] = df3['comprimento_nome_produto'].astype(int)
df3['comprimento_descricao_produto'] = df3['comprimento_descricao_produto'].astype(int)
df3['qtd_fotos_produto'] = df3['qtd_fotos_produto'].astype(int)
```

```python
df3.columns = df3.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
print("\n---- Nulos por coluna ----")
print(df_unido_fim.isna().sum())

print("\n---- Porcentagem de Nulos ----")
print(df_unido_fim.isna().mean() * 100)
```

```python
print("\n---- Duplicatas por colunas ----")
print(df_unido_fim.duplicated().sum())
```

O dataframe unido manteve alguns dos erros como tipos de colunas errados e valores nulos, então agora é necessário consertá-los novamente

```python
df_unido_fim['id_produto'] = df_unido_fim['id_produto'].fillna('Não identificado')
df_unido_fim['nome_categoria_produto'] = df_unido_fim['nome_categoria_produto'].fillna("Não classificado")
df_unido_fim['comprimento_nome_produto'] = df_unido_fim['comprimento_nome_produto'].fillna(df_unido_fim['comprimento_nome_produto'].median())
df_unido_fim['comprimento_descricao_produto'] = df_unido_fim['comprimento_descricao_produto'].fillna(df_unido_fim['comprimento_descricao_produto'].median())
df_unido_fim['qtd_fotos_produto'] = df_unido_fim['qtd_fotos_produto'].fillna(0)
df_unido_fim['peso_produto_g'] = df_unido_fim['peso_produto_g'].fillna(df_unido_fim['peso_produto_g'].median())
df_unido_fim['comprimento_produto_cm'] = df_unido_fim['comprimento_produto_cm'].fillna(df_unido_fim['comprimento_produto_cm'].median())
df_unido_fim['altura_produto_cm'] = df_unido_fim['altura_produto_cm'].fillna(df_unido_fim['altura_produto_cm'].median())
df_unido_fim['largura_produto_cm'] = df_unido_fim['largura_produto_cm'].fillna(df_unido_fim['largura_produto_cm'].median())
df_unido_fim['titulo_comentario_avaliacao'] = df_unido_fim['titulo_comentario_avaliacao'].fillna('Vazio')
df_unido_fim['mensagem_comentario_avaliacao'] = df_unido_fim['mensagem_comentario_avaliacao'].fillna('Vazio')
df_unido_fim['nota_avaliacao'] = df_unido_fim['nota_avaliacao'].fillna(0)
df_unido_fim['id_avaliacao'] = df_unido_fim['id_avaliacao'].fillna('Não identificado')
df_unido_fim['data_entrega_pedido_cliente'] = df_unido_fim['data_entrega_pedido_cliente'].ffill()
df_unido_fim['data_entrega_pedido_transporadora'] = df_unido_fim['data_entrega_pedido_transporadora'].ffill()
df_unido_fim['marca_temporal_compra_pedido'] = df_unido_fim['marca_temporal_compra_pedido'].ffill()
df_unido_fim['data_criacao_avaliacao'] = df_unido_fim['data_criacao_avaliacao'].ffill()
df_unido_fim['pedido_aprovado_as'] = df_unido_fim['pedido_aprovado_as'].ffill()
```

```python
df_unido_fim['marca_temporal_compra_pedido'] = pd.to_datetime(df_unido_fim['marca_temporal_compra_pedido'], errors='coerce')
df_unido_fim['pedido_aprovado_as'] = pd.to_datetime(df_unido_fim['pedido_aprovado_as'], errors='coerce')
df_unido_fim['data_entrega_pedido_transporadora'] = pd.to_datetime(df_unido_fim['data_entrega_pedido_transporadora'], errors='coerce')
df_unido_fim['data_entrega_pedido_cliente'] = pd.to_datetime(df_unido_fim['data_entrega_pedido_cliente'], errors='coerce')
df_unido_fim['data_estimada_entrega_pedido'] = pd.to_datetime(df_unido_fim['data_estimada_entrega_pedido'], errors='coerce')
df_unido_fim['data_limite_envio'] = pd.to_datetime(df_unido_fim['data_limite_envio'], errors='coerce')
df_unido_fim['comprimento_nome_produto'] = df_unido_fim['comprimento_nome_produto'].astype(int)
df_unido_fim['comprimento_descricao_produto'] = df_unido_fim['comprimento_descricao_produto'].astype(int)
df_unido_fim['qtd_fotos_produto'] = df_unido_fim['qtd_fotos_produto'].astype(int)
```
